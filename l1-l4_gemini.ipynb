## 1. Vyučovací hodina: Základní výpočty, proměnné a datové typy

**Motivace:** Výpočet základních geometrických a výkonových parametrů motoru.

```python
import math

# Zadané parametry motoru
vrtani_mm = 81.0      
zdvih_mm = 95.5       
pocet_valcu = 4
otacky = 3000         
p_e = 1.2e6           # Efektivní tlak [Pa]

# Převod jednotek
vrtani_m = vrtani_mm / 1000
zdvih_m = zdvih_mm / 1000

# Výpočet zdvihového objemu jednoho válce a celého motoru
V_valec = (math.pi * (vrtani_m ** 2) / 4) * zdvih_m
V_motor = V_valec * pocet_valcu

# Výpočet efektivního výkonu (čtyřdobý motor)
P_e = (p_e * V_motor * otacky) / (2 * 60)

print(f"Zdvihový objem: {V_motor*1000:.2f} litrů")
print(f"Efektivní výkon: {P_e/1000:.1f} kW")
```

## 2. Vyučovací hodina: Logické operace a rozhodování (If/Else)

**Motivace:** Motor se chová jinak na základě typu spalování a složení směsi.

```python
# --- Příklad 1: Hodnocení směsi ---
lambda_smes = 1.0

if lambda_smes == 1.0:
    spalovani_idealni = True
    print("Stechiometrická směs (ideální pro třícestný katalyzátor)")
elif lambda_smes > 1.0:
    spalovani_idealni = False
    print("Chudá směs (přebytek vzduchu, nižší spotřeba, hrozí NOx)")
else:
    spalovani_idealni = False
    print("Bohatá směs (nedostatek vzduchu, vyšší výkon, hrozí CO)")

print("\n")

# --- Příklad 2: Volba výpočtu teoretického oběhu ---
typ_motoru = "vznetovy" # možnosti: "zazehovy", "vznetovy"

if typ_motoru == "vznetovy":
    print("Spouštím výpočet Dieselova oběhu (izobarický přívod tepla)...")
    # Zde by následovaly specifické rovnice
elif typ_motoru == "zazehovy":
    print("Spouštím výpočet Ottova oběhu (izochorický přívod tepla)...")
else:
    print("Neznámý typ motoru!")
```

## 4. Vyučovací hodina: Vizualizace (Matplotlib)

**Motivace:** Vykreslení p-V diagramu vypočítaného Dieselova oběhu.

```python
import matplotlib.pyplot as plt

# Data z předchozí hodiny (zjednodušený tvar oběhu tvořený úsečkami)
objemy_l = [0.5, 0.03, 0.05, 0.5, 0.5]      # Převod na litry pro hezčí osu X
tlaky_bar = [1.0, 45.0, 45.0, 3.5, 1.0]     # Převod na bary pro hezčí osu Y

# Vytvoření grafu
plt.figure(figsize=(8, 5))
plt.plot(objemy_l, tlaky_bar, marker='o', color='red', label='Dieselův oběh', linewidth=2)

# Popisky bodů
body = ['1', '2', '3', '4', '1']
for i in range(4):
    plt.annotate(f"Bod {body[i]}", (objemy_l[i], tlaky_bar[i]), textcoords="offset points", xytext=(5,5))

# Formátování grafu
plt.title("p-V diagram teoretického oběhu vznětového motoru", fontsize=14)
plt.xlabel("Objem V [l]")
plt.ylabel("Tlak p [bar]")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Vykreslení na obrazovku
plt.show()
```
