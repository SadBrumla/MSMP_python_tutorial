## Část 1: Definice proměnných a datové typy

V Pythonu nemusíme deklarovat typ proměnné předem. Jazyk si jej odvodí sám podle toho, jakou hodnotu do proměnné přiřadíme pomocí rovnítka `=`. 

Pro inženýrské výpočty spalovacích motorů budeme nejčastěji používat:
- `int` (celá čísla) – např. počet válců, počet ventilů.
- `float` (desetinná čísla, píšeme s tečkou!) – např. vrtání, zdvih, kompresní poměr.
- `str` (textové řetězce, píšeme do uvozovek) – např. označení motoru, typ paliva.

Důležité pravidlo: Názvy proměnných píšeme bez mezer, ideálně tzv. `snake_case` (např. `pocet_valcu`).

```python
# VZOROVÝ PŘÍKLAD 1
# Definice parametrů legendárního motoru 1.9 TDI

oznaceni_motoru = "1.9 TDI PD"  # Datový typ: str (string)
pocet_valcu = 4                 # Datový typ: int (integer)
vrtani_mm = 79.5                # Datový typ: float (floating point number)
zdvih_mm = 95.5                 # Datový typ: float
preplnovani = True              # Datový typ: bool (boolean - Pravda/Nepravda)

# Funkce type() nám prozradí, jaký datový typ Python proměnné přiřadil
print(type(vrtani_mm))
print(type(oznaceni_motoru))
```

## Část 2: Matematické operace a převody jednotek

Python funguje jako velmi mocná kalkulačka. Základní operátory jsou:
- `+` sčítání
- `-` odčítání
- `*` násobení
- `/` dělení
- `**` mocnina (pozor, v Pythonu se nepoužívá `^`!)

Pro složitější matematiku (např. Ludolfovo číslo $\pi$) musíme naimportovat vestavěnou knihovnu `math`.

*Tip pro strojaře: Vždy si hlídejte jednotky! Pro výpočty objemu nebo termodynamiky je nejlepší hned na začátku převést milimetry na metry (Základní jednotky SI).* 

```python
# VZOROVÝ PŘÍKLAD 2
# Výpočet zdvihového objemu motoru 1.9 TDI z předchozí části

import math  # Načtení matematické knihovny pro použití math.pi

# 1. Převod na metry
vrtani_m = vrtani_mm / 1000
zdvih_m = zdvih_mm / 1000

# 2. Výpočet plochy pístu (S = pi * r^2 nebo pi * d^2 / 4)
plocha_pistu_m2 = (math.pi * (vrtani_m ** 2)) / 4

# 3. Výpočet objemu jednoho válce (V = S * z) a celého motoru
objem_valce_m3 = plocha_pistu_m2 * zdvih_m
objem_motoru_m3 = objem_valce_m3 * pocet_valcu

# 4. Převod na litry pro lepší představu (1 m3 = 1000 l)
objem_motoru_litry = objem_motoru_m3 * 1000

print("Zdvihový objem v litrech:")
print(objem_motoru_litry)
```

Nyní spojíme vše, co jsme se dnes naučili.
Čistý Dieselův oběh (teplo přiváděné pouze při konstantním tlaku) se u moderních motorů příliš nepoužívá. Reálnému průběhu tlaku ve válci se mnohem více blíží **Sabatheho oběh (Smíšený oběh / Dual cycle)**.
Teplo je zde přiváděno ve dvou fázích: část izochoricky (konst. objem - prudký nárůst tlaku po vznícení) a část izobaricky (konst. tlak - hoření během pohybu pístu dolů).

**Vaším úkolem je dopočítat stavové veličiny (tlak $p$, teplotu $T$, objem $v$) pro body 1 až 5 tohoto oběhu.**

**Stručný algoritmus oběhu:**
- **Bod 1:** Sání (zadáno).
- **Bod 2 (Izoentropická komprese):** $v_2 = v_1 / \epsilon$, $p_2 = p_1 \cdot \epsilon^\kappa$, $T_2 = T_1 \cdot \epsilon^{\kappa-1}$
- **Bod 3 (Izochorický přívod tepla):** $v_3 = v_2$, $p_3 = p_2 \cdot \alpha$, $T_3 = T_2 \cdot \alpha$
- **Bod 4 (Izobarický přívod tepla):** $p_4 = p_3$, $v_4 = v_3 \cdot \beta$, $T_4 = T_3 \cdot \beta$
- **Bod 5 (Izoentropická expanze):** $v_5 = v_1$, $p_5 = p_4 \cdot (v_4 / v_5)^\kappa$, $T_5 = T_4 \cdot (v_4 / v_5)^{\kappa-1}$
