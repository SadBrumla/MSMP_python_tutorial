### Datové typy

```python
# Komentář - jen pro programátora, nemá vliv na kód

jmeno = "Octavia"       # str (string)
pocet_valcu = 4         # int (inteeger)
vrtani = 81.0           # float 
zdvih = 95.5            # float 
vznetovy = True         # bool
```

S těmito proměnnými je možné provádět operace (matematické, logické)

```python
vrtani * zdvih
```

jejich výsledky ukládat do dalších promměných (pomocí =) a zobrazovat pomocí print()

```python
objem = vrtani * zdvih
print(objem)
```

Typ proměnné lze ověřit pomocí zavolání funkce type() a zobrazit pomocí print

### OTÁZKA?

```python
# jaky bude výstup těchto promměných?
type(jmeno)
type(pocet_valcu)
type(vznetovy)
type(objem)
```
