## Shrnutí: Logické operátory a Boolean
Datový typ **Boolean** (zkráceně `bool`) může nabývat pouze dvou hodnot: `True` (Pravda) nebo `False` (Nepravda). Dáváme si pozor na velká počáteční písmena!

**Základní srovnávací operátory:**
- `==` Rovná se (porovnání)
- `!=` Nerovná se
- `>`, `<` Větší než, menší než
- `>=`, `<=` Větší nebo rovno, menší nebo rovno

**Logické operátory (spojování podmínek):**
- `and` (a zároveň) - obě podmínky musí platit
- `or` (nebo) - stačí, když platí alespoň jedna
- `not` (negace) - obrátí True na False a naopak

## Část 3: Cyklus While (Dokud)
Cyklus `while` opakuje blok kódu neustále dokola, *dokud* je jeho podmínka `True`.

````python
# VZOROVÝ PŘÍKLAD 3: Zahřívání motoru
teplota_oleje = 20
provozni_teplota = 90

print("Začínáme zahřívat motor...")
while teplota_oleje < provozni_teplota:
    teplota_oleje += 15  # Toto je zkratka pro: teplota_oleje = teplota_oleje + 15
    print(f"Aktuální teplota: {teplota_oleje} °C")

print("Motor dosáhl provozní teploty. Můžete jet naplno!")
````

## Část 4: Seznamy (Lists) a Cyklus For
Seznam nám umožňuje uložit více hodnot do jedné proměnné. Cyklus `for` pak používáme k projití každého prvku v seznamu jeden po druhém.

````python
# VZOROVÝ PŘÍKLAD 4: Zpracování seznamu otáček
otacky_motoru = [1000, 2500, 4000, 6000]

print("Výpočet střední pístové rychlosti (zdvih = 0.08 m):")
zdvih = 0.08

# Pro každý prvek (pojmenovaný 'ot') v seznamu 'otacky_motoru':
for ot in otacky_motoru:
    c_s = 2 * zdvih * (ot / 60)
    print(f"Otáčky: {ot} ot/min -> c_s = {c_s:.1f} m/s")
````

## Bonus: Novinka z Pythonu 3.10 - Příkaz Match / Case (Switch)
Pokud potřebujeme testovat jednu proměnnou vůči mnoha konkrétním hodnotám (dříve se používaly dlouhé řetězce `if` - `elif` - `elif`), můžeme nově využít `match` (v jiných jazycích známé jako `switch`).

````python
oznaceni = "TDI"

match oznaceni:
    case "TSI":
        print("Turbocharged Stratified Injection (Zážehový, přímý vstřik)")
    case "TDI":
        print("Turbocharged Direct Injection (Vznětový, přímý vstřik)")
    case "MPI":
        print("Multi Point Injection (Zážehový, nepřímý vstřik)")
    case _:
        print("Neznámé označení motoru!")
````
